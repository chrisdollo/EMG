# Efficient (8-channel)

Same runners as `baseline.ipynb`, restricted to 8 channels (and 8 per-channel
features for the feature-based path). Uses the clustering selection if present,
otherwise falls back to a random selection.

Runs locally or on Colab — the first cell auto-mounts Drive and `cd`s into the
project when on Colab (edit `BASE` there if your Drive layout differs).

```python
import os, sys

# On Colab: mount Drive and cd into the project so the relative paths below work.
# Locally this block is skipped. Adjust BASE if your Drive layout differs.
try:
    from google.colab import drive
    drive.mount('/content/drive')
    os.chdir('/content/drive/Othercomputers/My Mac/putEMG prime')
    print('Colab: cwd =', os.getcwd())
except ModuleNotFoundError:
    pass

import numpy as np
sys.path.insert(0, 'src')
import runners

# Pick 8 channels and 8 per-channel features — from clustering if available, else random
rng = np.random.default_rng(42)

ch_path = 'clustering & analysis/feat_representative_channels.npy'
if os.path.exists(ch_path):
    channels = np.load(ch_path)
    print(f'channels    : loaded {len(channels)} from {ch_path} -> {channels.tolist()}')
else:
    channels = np.sort(rng.choice(24, 8, replace=False))
    print(f'channels    : RANDOM (no clustering output) -> {channels.tolist()}')

feat_path = 'clustering & analysis/feat_representative_features.npy'
if os.path.exists(feat_path):
    feature_idx = np.load(feat_path)
    print(f'feature_idx : loaded {len(feature_idx)} from {feat_path} -> {feature_idx.tolist()}')
else:
    feature_idx = np.sort(rng.choice(41, 8, replace=False))
    print(f'feature_idx : RANDOM (no selection) -> {feature_idx.tolist()}')
```

```python
# Within-subject — deep learning (8 channels)
runners.run_within(
    type='deep_learning',
    FORCE_RERUN=False,
    input_folder='data/processed gestures',
    output_weight_folder='weights/efficient/within_dl',
    result_folder='results/efficient',
    experiment='efficient',
    channels=channels,
)
```

```python
# Within-subject — feature-based (8 channels × 8 features)
runners.run_within(
    type='feature_based',
    FORCE_RERUN=False,
    input_folder='data/features',
    output_weight_folder='weights/efficient/within_feat',
    result_folder='results/efficient',
    experiment='efficient',
    channels=channels, feature_idx=feature_idx,
)
```

```python
# Cross-subject LOSO — deep learning (8 channels)
runners.run_cross(
    type='deep_learning',
    FORCE_RERUN=False,
    input_folder='data/processed gestures',
    output_weight_folder='weights/efficient/cross_dl',
    result_folder='results/efficient',
    experiment='efficient',
    channels=channels,
)
```

```python
# Cross-subject LOSO — feature-based (8 channels × 8 features)
runners.run_cross(
    type='feature_based',
    FORCE_RERUN=False,
    input_folder='data/features',
    output_weight_folder='weights/efficient/cross_feat',
    result_folder='results/efficient',
    experiment='efficient',
    channels=channels, feature_idx=feature_idx,
)
```

```python
# Cross-subject LOSO — feature-based (8 channels × 8 features)
runners.run_cross(
    type='feature_based',
    FORCE_RERUN=False,
    input_folder='data/features',
    output_weight_folder='weights/efficient/cross_feat',
    result_folder='results/efficient',
    channels=channels, feature_idx=feature_idx,
)
```
